# Dual-Target Grounded-Execute ChartQA — Colab runner

Same package, same CLI, different host. Nothing here is Colab-specific except
how secrets are read and where the repo is cloned.

**Before running:** add `GITHUB_TOKEN`, `GITHUB_USER`, `HF_TOKEN` (and optionally
`WANDB_API_KEY`) to the 🔑 *Secrets* panel on the left, with notebook access on.

**Runtime → Change runtime type → T4 GPU.**


```python
from google.colab import userdata
import os, subprocess, sys

for k in ("GITHUB_TOKEN", "GITHUB_USER", "HF_TOKEN", "WANDB_API_KEY"):
    try:
        os.environ[k] = userdata.get(k)
    except Exception:
        print(f"secret {k} not set (continuing)")
```

```python
REPO = f"https://{os.environ['GITHUB_TOKEN']}@github.com/{os.environ['GITHUB_USER']}/chartqa-dual-target.git"
subprocess.run(["git", "clone", "--depth", "1", "-b", "main", REPO, "/content/repo"], check=True)
```

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", "/content/repo[gpu,eval]"], check=True)
```

```python
subprocess.run([sys.executable, "-c",
                "from chartqa_dt.env import get_env; print(get_env().describe())"], check=True)
```

```python
CMD = "cdt-train --stage smoke --config configs/model_qwen3vl2b.yaml"
subprocess.run(CMD, shell=True, cwd="/content/repo", check=True)
```
